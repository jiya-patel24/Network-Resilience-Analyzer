```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
student_data = {
    "attendance_pct": [92, 68, 45, 88, 52, 33, 76, 48, 60, 85],
    "assignment_score": [96, 72, 38, 90, 58, 28, 82, 42, 65, 88],
    "lms_activity": [22, 14, 6, 19, 9, 4, 16, 7, 11, 18],
    "internal_marks": [87, 62, 40, 83, 57, 32, 78, 45, 61, 84],
    "participation_level": [9, 6, 3, 8, 5, 2, 7, 4, 6, 8],
    "risk_label": [0, 1, 2, 0, 1, 2, 0, 1, 1, 0]
}

df = pd.DataFrame(student_data)
df
```

```python
X = df.drop("risk_label", axis=1)
y = df["risk_label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1
)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=1
)

rf_model.fit(X_train, y_train)
```

```python
predictions = rf_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, predictions))
print("\nClassification Report:\n")
print(classification_report(y_test, predictions))
```

```python
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
}).sort_values(by="Importance", ascending=False)

importance
```

```python
sample_student = pd.DataFrame([{
    "attendance_pct": 50,
    "assignment_score": 48,
    "lms_activity": 6,
    "internal_marks": 45,
    "participation_level": 4
}])

risk_prediction = rf_model.predict(sample_student)[0]

risk_names = {0: "Low Risk", 1: "Medium Risk", 2: "High Risk"}
print("Predicted Dropout Risk:", risk_names[risk_prediction])
```
